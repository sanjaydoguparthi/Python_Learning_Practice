```python
# Program to perform all arithmetic operations

a = 10
b = 5

addition = a + b
subtraction = a - b
multiplication = a * b
division = a / b
modulus = a % b
exponentiation = a ** b
floor_division = a // b

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Modulus:", modulus)
print("Exponentiation:", exponentiation)
print("Floor Division:", floor_division)
```

```python
# Examples of valid and invalid identifiers in Python

# Valid identifiers
my_var = 10
_var2 = 20
Var3 = 30
myVar = 40

print("Valid identifiers:")
print("my_var =", my_var)
print("_var2 =", _var2)
print("Var3 =", Var3)
print("myVar =", myVar)

# Invalid identifiers (uncommenting these lines will cause a SyntaxError)
# 2var = 50      # Cannot start with a digit
# my-var = 60    # Hyphens are not allowed
# for = 70       # 'for' is a reserved keyword

print("\nIdentifiers must start with a letter or underscore, cannot contain spaces or hyphens, and cannot be a reserved keyword.")
```

```python
# Program to demonstrate relational operators

print("a =", a)
print("b =", b)

print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
print("a != b:", a != b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
```

```python

# Program to demonstrate list operations
my_list = [1, 2, 3, 4, 5]
print("\nList Operations:")
print("Original list:", my_list)        
print("Length of list:", len(my_list))
print("First element:", my_list[0])
print("Last element:", my_list[-1])
my_list.append(6)   
print("List after appending 6:", my_list)
my_list.remove(3)
print("List after removing 3:", my_list)
print("List after sorting:", sorted(my_list))
print("List after reversing:", my_list[::-1])
# Program to demonstrate list operations
print("List after clearing:", my_list.clear())
print("List after clearing:", my_list)
```
