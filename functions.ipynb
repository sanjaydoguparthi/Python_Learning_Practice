```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
# call this function
even_or_odd(20)
```

```python
### Variable Length Arguments
## Positional And Keywords arguments

def print_numbers(*sanjay):
    for number in sanjay:
        print(number)
```

```python
## call this function
print_numbers(1,2,4,5,3,6,'sanjay')
```

```python
## Positional arguments
def print_numbers_1(*args):
    for number in args:
        print(number)
```

```python
# call this functi
print_numbers_1(22,55,41, 1 )
```

```python
### Keywords Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name = 'sanjay', age = 24, city = 'chirala')
```

```python
## Positional arguments and  Keywords Arguments

def print_details_1(*args,**kwargs):
    for value in args:
        print(f" Positional arument :{value}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details_1(1,2,3, NAME = 'sanjay', AGE = 24, CITY = 'chirala', STATE = 'andhra pradesh')
```
